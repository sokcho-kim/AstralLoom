```python
# Explicitly load scheduler first
import torch
from diffusers import DDIMScheduler, DiffusionPipeline

# Load scheduler separately
scheduler = DDIMScheduler.from_pretrained(
    "VAST-AI/TripoSG", 
    subfolder="scheduler",
    config_file_name="scheduler_config.json"  # Explicitly specify
)

# Then load pipeline with custom scheduler
pipeline = DiffusionPipeline.from_pretrained(
    "VAST-AI/TripoSG",
    scheduler=scheduler,
    torch_dtype=torch.float16,
    use_safetensors=True
)
```
